```python
print("\n missing values:\n")
missing_values = df.isnull().sum()
print(missing_values[missing_values > 0])
print(f"\nDuplicate Rows:{df.duplicated().sum()}")
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d-%m-%Y')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format='%d-%m-%Y')

print("\nDate columns converted:\n")
print(f"Order Date: {df['Order Date'].dtype}")
print(f"Ship Date: {df['Ship Date'].dtype}")
```

```python
print("\n======== Unique values in categorical columns ========\n")
categorical_columns=['Customer Segment','Category','Region','Payment Mode','Delivery Status','Auto Reorder']
for column in categorical_columns:
    print(f"\n{column}:\n")
    print(df[column].value_counts())

    plt.figure(figsize=(6,6))
    df[column].value_counts().plot(kind='bar',color='skyblue')
    plt.title(f'Distribution of {column}')
    plt.xticks(rotation=30)
    plt.tight_layout()
    plt.show()
```

```python
print("\n======== Summary statistics for numerical columns ========\n")
numerical_columns = ['Quantity','Unit Price','Discount (%)','Sales Amount','Cost Price','Profit','Stock Left','Reorder Quantity']
for column in numerical_columns:
    print(f"\n{column}:\n")
    print(f"Mean: {df[column].mean():.2f}")
    print(f"Median: {df[column].median():.2f}")
    print(f"Standard Deviation: {df[column].std():.2f}")
    print(f"Minimum: {df[column].min():.2f}")
    print(f"Maximum: {df[column].max():.2f}")
    plt.figure(figsize=(8, 4))
    plt.subplot(1, 2, 1)
    df[column].hist(bins=30, color='lightgreen', alpha=0.7)
    plt.title(f'Distribution of {column}')
    plt.subplot(1, 2, 2)
    df.boxplot(column=column)
    plt.title(f'Boxplot of {column}')
    plt.tight_layout()
    plt.show()
```

```python
print("\n======== Category Performance ========\n")
category_performance=df.groupby('Category').agg({'Sales Amount':'sum','Profit':'sum','Order ID':'count'}).rename(columns={'Order ID':'Order Count'})
category_performance['Profit Margin']=(category_performance['Profit']/category_performance['Sales Amount'])*100
print(category_performance,'\n\n')
fig,axis=plt.subplots(2,2,figsize=(15,12))

category_performance['Sales Amount'].plot(kind='bar',ax=axis[0,0],color='blue',alpha=0.7)
axis[0,0].set_title('Total Sales by Category')
axis[0,0].tick_params(axis='x', rotation=30)

category_performance['Profit'].plot(kind='bar',ax=axis[0,1],color='green',alpha=0.7)
axis[0,1].set_title('Total Profit by Category')
axis[0,1].tick_params(axis='x', rotation=30)

category_performance['Profit Margin'].plot(kind='bar',ax=axis[1,0],color='orange',alpha=0.7)
axis[1,0].set_title('Profit Margin by Category (%)')
axis[1,0].tick_params(axis='x', rotation=30)

category_performance['Order Count'].plot(kind='bar',ax=axis[1,1],color='red',alpha=0.7)
axis[1,1].set_title('Total Orders by Category')
axis[1,1].tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.show()
```

```python
print("\n======== Regional Performance ========\n")
regional_performance=df.groupby('Region').agg({'Sales Amount':'sum','Profit':'sum','Order ID':'count'}).rename(columns={'Order ID':'Order Count'})
regional_performance['Profit Margin']=(regional_performance['Profit']/regional_performance['Sales Amount'])*100
print(regional_performance,'\n\n')

plt.figure(figsize=(12,8))
plt.subplot(2,2,1)
regional_performance['Sales Amount'].plot(kind='pie',autopct='%1.1f%%')
plt.title('Sales Distribution by Region')

plt.subplot(2,2,2)
regional_performance['Profit'].plot(kind='pie',autopct='%1.1f%%')
plt.title('Profit Distribution by Region')

plt.subplot(2,2,3)
sns.barplot(x=regional_performance.index,y='Order Count',data=regional_performance)
plt.title('Order Count by Region')
plt.xticks(rotation=30)

plt.subplot(2,2,4)
sns.barplot(x=regional_performance.index,y='Profit Margin',data=regional_performance)
plt.title('Profit Margin by Region (%)')
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```

```python
print("\n======== Time Series Analysis ========\n")

df['Order Month'] = df['Order Date'].dt.to_period('M')
df['Order Year'] = df['Order Date'].dt.year

monthly_trends=df.groupby('Order Month').agg({'Sales Amount':sum,'Profit':'sum','Order ID':'count'}).rename(columns={'Order ID':'Order Count'})

print("Monthly Trends:")
print(monthly_trends.tail(10),'\n\n')

plt.figure(figsize=(15,10))

plt.subplot(3,1,1)
monthly_trends['Sales Amount'].plot(kind='line',color='blue',marker='o')
plt.title('Monthly Sales Trend')
plt.ylabel('Sales Amount')
plt.grid(True)

plt.subplot(3,1,2)
monthly_trends['Profit'].plot(kind='line',color='green',marker='o')
plt.title('Monthly Profit Trend')
plt.ylabel('Profit')
plt.grid(True)

plt.subplot(3,1,3)
monthly_trends['Order Count'].plot(kind='line',color='red',marker='o')
plt.title('Monthly Order Count Trend')
plt.ylabel('Order Count')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
print("\n======== Correlation Analysis ========\n")

numerical_df = df[['Quantity', 'Unit Price', 'Discount (%)', 'Sales Amount', 'Cost Price', 'Profit']]

correlation_matrix = numerical_df.corr()
print("Correlation Matrix:")
print(correlation_matrix, '\n\n')

plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0, square=True, linewidths=0.5)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
print("\n======== Customer Segment Analysis ========\n")

Segment_analaysis=df.groupby('Customer Segment').agg({'Sales Amount':['sum','mean'],'Profit':['sum','mean'],'Order ID':'count','Discount (%)': 'mean'}).round(2)

numerical_df = df[['Quantity', 'Unit Price', 'Discount (%)', 'Sales Amount', 'Cost Price', 'Profit']]

Segment_analaysis.columns=['Total Sales','Avg Sales','Total Profit','Avg Profit','Order Count','Avg Discount']
print(Segment_analaysis,'\n\n')

fig,axis=plt.subplots(2,2,figsize=(15,10))

Segment_analaysis['Total Sales'].plot(kind='bar',ax=axis[0,0],color='lightblue')
axis[0,0].set_title('Total Sales by Customer Segment')
axis[0,0].tick_params(axis='x',rotation=30)

Segment_analaysis['Total Profit'].plot(kind='bar',ax=axis[0,1],color='lightgreen')
axis[0,1].set_title('Total Profit by Customer Segment')
axis[0,1].tick_params(axis='x',rotation=30)

Segment_analaysis['Avg Profit'].plot(kind='bar',ax=axis[1,0],color='Orange')
axis[1,0].set_title('Average Profit by Customer Segment')
axis[1,0].tick_params(axis='x',rotation=30)

Segment_analaysis['Avg Discount'].plot(kind='bar',ax=axis[1,1],color='red')
axis[1,1].set_title('Average Discount by Customer Segment')
axis[1,1].tick_params(axis='x',rotation=30)

plt.tight_layout()
plt.show()
```

```python
print("\n ==== key insights summary ====\n")

# overall metrices
total_sales=df['Sales Amount'].sum()
total_profit=df['Profit'].sum()
total_orders=df['Order ID'].nunique()
avg_profit_margin=(total_profit/total_sales)*100

print(f"overall performance:\n")
print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Total Orders: ${total_orders}")
print(f"Average Profit Margin: ${avg_profit_margin:.2f}%")

# top performance categories
top_categories=df.groupby('Category')['Profit'].sum().sort_values(ascending=False)
print(f"\nTop Performance Categories:\n")
for i ,(category,profit) in enumerate(top_categories.items(),1):
    print(f"{i}.category: ${profit:,.2f}")

# regional performance
best_region=df.groupby('Region')['Profit'].sum().idxmax()
print(f"\nBest Performance Region: {best_region}")

# delivery performance
delivery_stats=df['Delivery Status'].value_counts(normalize=True)*100
print(f"\nDelivery Status:")
for status,percentage in delivery_stats.items():
    print(f"{status}: {percentage:.1f}%")

# customer segments
best_segment=df.groupby('Customer Segment')['Profit'].sum().idxmax()
print(f"\nMost Profitable Customer Segment: {best_segment}")
```

```python
analysis_results = {
    'total_sales': total_sales,
    'total_profit': total_profit,
    'total_orders': total_orders,
    'avg_profit_margin': avg_profit_margin,
    'top_categories': top_categories,
    'best_region': best_region,
    'best_segment': best_segment
}

# Convert to DataFrame and save
results_df = pd.DataFrame([analysis_results])

results_df.to_csv('eda_analysis_summary.csv', index=False)

print("\nAnalysis Summary Saved to 'eda_analysis_summary.csv'")
```
